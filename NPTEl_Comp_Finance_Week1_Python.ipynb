## The Bond Pricer in Pseudocode

```
function bond_price(cash_flows, times, r):
    total <- 0
    for each (CF, t) in (cash_flows, times):
        discount_factor <- 1 / (1 + r) ^ t
        total <- total + CF * discount_factor
    return total
```

````python
def bond_price(cash_flows, times, r):
    total = 0
    for cf, t in zip(cash_flows, times):
        discount_factor = 1 / (1 + r) ** t
        total += cf * discount_factor
    return total

# Example usage (optional, can be removed if not needed for the final solution):
cash_flows = [100, 100, 1100]
times = [1, 2, 3]
r = 0.05
price = bond_price(cash_flows, times, r)
print(f"Bond Price: {price:.2f}")
````

````python
cash_flows = [25, 25, 25, 25, 1025]
times = [1, 2, 3, 4, 5]
r = 0.02
price = bond_price(cash_flows, times, r)
print(f"Bond Price: {price:.2f}")
````

 ## Example: Rs 100 at 8\% nominal rate after one year

Convention  | Formula                      | Value.     |
------------|------------------------------|------------|

### Implementing Compounding Conventions

Let's implement the different compounding formulas shown in your slide using Python. We'll start by defining the initial principal, nominal interest rate, and time period.

````python
# Define the parameters from the example
principal = 100
nominal_rate = 0.08  # 8%
time_years = 1
````

````python
import math

# Annual Compounding
annual_fv = principal * (1 + nominal_rate) ** time_years

# Semi-annual Compounding
semi_annual_compounds = 2
semi_annual_rate = nominal_rate / semi_annual_compounds
semi_annual_fv = principal * (1 + semi_annual_rate) ** (semi_annual_compounds * time_years)

# Monthly Compounding
monthly_compounds = 12
monthly_rate = nominal_rate / monthly_compounds
monthly_fv = principal * (1 + monthly_rate) ** (monthly_compounds * time_years)

# Daily Compounding
daily_compounds = 365
daily_rate = nominal_rate / daily_compounds
daily_fv = principal * (1 + daily_rate) ** (daily_compounds * time_years)

# Continuous Compounding
continuous_fv = principal * math.exp(nominal_rate * time_years)

# Store results in a dictionary
results = {
    'Convention': [
        'Annual',
        'Semi-annual',
        'Monthly',
        'Daily',
        'Continuous'
    ],
    'Formula Example': [
        '100 * (1.08)^1',
        '100 * (1.04)^2',
        '100 * (1 + 0.08/12)^12',
        '100 * (1 + 0.08/365)^365',
        '100 * e^(0.08)'
    ],
    'Value': [
        annual_fv,
        semi_annual_fv,
        monthly_fv,
        daily_fv,
        continuous_fv
    ]
}

# Convert to DataFrame for display
import pandas as pd
df_results = pd.DataFrame(results)
display(df_results.round(2))
````

## Algorithm: Computing $e^x$ via Taylor Series

```
function exp_approx(x, tolerance):
    total <- 1
    term  <- 1
    n     <- 1
    while |term| > tolerance:
        term  <- term * x / n
        total <- total + term
        n     <- n + 1
    return total
```

````python
import math

def exp_approx(x, tolerance):
    total = 1.0
    term = 1.0
    n = 1

    while abs(term) > tolerance:
        term = term * x / n
        total += term
        n += 1

    return total

# Example usage:
print(exp_approx(2.0, 1e-7))   # Output: ~7.38905609893
print(math.exp(2.0))            # Built-in e^2 for comparison
````

````python
import time
# Input values
x = 2.0
tolerance = 1e-9

# Time for exp_approx
start_approx = time.perf_counter()
res_approx = exp_approx(x, tolerance)
end_approx = time.perf_counter()
time_approx = end_approx - start_approx

# Time for math.exp
start_builtin = time.perf_counter()
res_builtin = math.exp(x)
end_builtin = time.perf_counter()
time_builtin = end_builtin - start_builtin

# Results
print(f"exp_approx result : {res_approx:.10f} | Time: {time_approx * 1e6:.3f} microseconds")
print(f"math.exp result   : {res_builtin:.10f} | Time: {time_builtin * 1e6:.3f} microseconds")
````
