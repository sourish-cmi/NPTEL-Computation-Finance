## The Bond Pricer in Pseudocode

```
function bond_price(cash_flows, times, r):
    total <- 0
    for each (CF, t) in (cash_flows, times):
        discount_factor <- 1 / (1 + r) ^ t
        total <- total + CF * discount_factor
    return total
```

````r
bond_price <- function(cash_flows, times, r) {
  discount_factors <- 1 / (1 + r)^times
  total <- sum(cash_flows * discount_factors)
  return(total)
}

# Example usage:
cash_flows <- c(100, 100, 1100)
times <- c(1, 2, 3)
r <- 0.05

price <- bond_price(cash_flows, times, r)
cat(sprintf("Bond Price: %.2f\n", price))
````

````r
# Parameters
cash_flows <- c(25, 25, 25, 25, 1025)
times <- c(1, 2, 3, 4, 5)
r <- 0.02

# Calculation
price <- bond_price(cash_flows, times, r)
cat(sprintf("Bond Price: %.2f\n", price))
````

### Implementing Compounding Conventions

Let's implement the different compounding formulas shown in your slide using Python. We'll start by defining the initial principal, nominal interest rate, and time period.

````r
# Parameters
principal <- 100
nominal_rate <- 0.08  # 8%
time_years <- 1
````

````r
# Parameters
principal <- 100
nominal_rate <- 0.08  # 8%
time_years <- 1

# Compounding calculations
annual_fv <- principal * (1 + nominal_rate)^time_years

semi_annual_compounds <- 2
semi_annual_rate <- nominal_rate / semi_annual_compounds
semi_annual_fv <- principal * (1 + semi_annual_rate)^(semi_annual_compounds * time_years)

monthly_compounds <- 12
monthly_rate <- nominal_rate / monthly_compounds
monthly_fv <- principal * (1 + monthly_rate)^(monthly_compounds * time_years)

daily_compounds <- 365
daily_rate <- nominal_rate / daily_compounds
daily_fv <- principal * (1 + daily_rate)^(daily_compounds * time_years)

continuous_fv <- principal * exp(nominal_rate * time_years)

# Create data frame
df_results <- data.frame(
  Convention = c('Annual', 'Semi-annual', 'Monthly', 'Daily', 'Continuous'),
  `Formula Example` = c('100 * (1.08)^1', '100 * (1.04)^2', '100 * (1 + 0.08/12)^12', '100 * (1 + 0.08/365)^365', '100 * e^(0.08)'),
  Value = round(c(annual_fv, semi_annual_fv, monthly_fv, daily_fv, continuous_fv), 2),
  check.names = FALSE
)

# Display results
print(df_results)
````

## Algorithm: Computing $e^x$ via Taylor Series

```
function exp_approx(x, tolerance):
    total <- 1
    term  <- 1
    n     <- 1
    while |term| > tolerance:
        term  <- term * x / n
        total <- total + term
        n     <- n + 1
    return total
```

````r
exp_approx <- function(x, tolerance) {
  total <- 1.0
  term <- 1.0
  n <- 1

  while (abs(term) > tolerance) {
    term <- term * x / n
    total <- total + term
    n <- n + 1
  }

  return(total)
}

# Example usage:
cat("exp_approx result:", exp_approx(2.0, 1e-7), "\n") # Output: ~7.389056
cat("exp result       :", exp(2.0), "\n")            # Built-in e^2 for comparison
````

````r
# Input values
x <- 2.0
tolerance <- 1e-9

# Time for exp_approx
start_approx <- Sys.time()
res_approx <- exp_approx(x, tolerance)
end_approx <- Sys.time()
time_approx <- as.numeric(end_approx - start_approx, units = "secs")

# Time for built-in exp()
start_builtin <- Sys.time()
res_builtin <- exp(x)
end_builtin <- Sys.time()
time_builtin <- as.numeric(end_builtin - start_builtin, units = "secs")

# Results
cat(sprintf("exp_approx result : %.10f | Time: %.3f microseconds\n",
            res_approx, time_approx * 1e6))
cat(sprintf("exp result        : %.10f | Time: %.3f microseconds\n",
            res_builtin, time_builtin * 1e6))
````
